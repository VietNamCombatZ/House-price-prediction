```python
from datasets import load_dataset
from sklearn.model_selection import train_test_split
import pandas as pd

# Load dataset
ds = load_dataset("tinixai/vietnam-real-estates")

# Convert sang pandas
df = ds["train"].to_pandas()

# Chỉ lấy dữ liệu nhà
df_house = df[df["property_type_name"] == "Nhà"]

# Kiểm tra
print(df_house.shape)

# Xem vài dòng
df_house.head(3)
```

```python
df_house = df_house.reset_index(drop=True)
df_house["property_type_name"].unique()
```

```python
missing_values = df_house.isnull().sum()

missing_percentage = (missing_values / len(df_house)) * 100

missing_df = pd.DataFrame({
    "Missing Values": missing_values,
    "Missing (%)": missing_percentage
})

missing_df.sort_values(by="Missing Values", ascending=False)
```
