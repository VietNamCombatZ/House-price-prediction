```python
from datasets import load_dataset
from sklearn.model_selection import train_test_split
import pandas as pd

# Load dataset
ds = load_dataset("tinixai/vietnam-real-estates")

# Convert sang pandas
df = ds["train"].to_pandas()

# Chỉ lấy dữ liệu nhà
df_house = df[df["property_type_name"] == "Nhà"]

# Kiểm tra
print(df_house.shape)

# Xem vài dòng
df_house.head()
```

```python
df_house = df_house.reset_index(drop=True)
df_house["property_type_name"].unique()
```

```python
df_house.head()
df_house.info()
df_house.columns
df_house.isnull().sum()
```

```python

df_sample = df_house.sample(
    n=300000,
    random_state=42
)
# df_sample = df.sample(
#     n=300000,
#     random_state=42
# )

train_df, test_df = train_test_split(
    df_sample,
    test_size=0.2,
    random_state=42
)
```

```python
train_df.to_csv(
    "../data/raw/raw_data_train.csv",
    index=False
)

test_df.to_csv(
    "../data/raw/raw_data_test.csv",
    index=False
)
```
